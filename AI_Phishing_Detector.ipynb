```python
import re
from sklearn.ensemble import RandomForestClassifier

def extract_features(url):
    return [
        len(url),
        url.count('.'),
        url.count('-'),
        url.count('@'),
        1 if url.startswith('https') else 0,
        1 if re.search(r'\d+\.\d+\.\d+\.\d+', url) else 0,
        1 if any(word in url.lower() for word in ['login', 'verify', 'bank', 'free', 'update', 'secure']) else 0
    ]

data = [
    ("https://google.com", 0),
    ("https://kaist.ac.kr", 0),
    ("https://github.com", 0),
    ("https://ibm.com", 0),
    ("http://free-prize-update-login.com", 1),
    ("http://192.168.1", 1),
    ("http://paypal-account-verify-now@security.net", 1),
    ("http://login-update-your-bank-details.xyz", 1)
]

X = [extract_features(url) for url, label in data]
y = [label for url, label in data]

model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(X, y)

test_urls = [
    "https://kaist.ac.kr",
    "http://free-iphone-winner-claim-now.info",
    "http://security-alert-verify-account.com"
]

print("--- AI PHISHING DETECTOR RESULTS ---")
for url in test_urls:
    features = [extract_features(url)]
    pred = model.predict(features)[0]
    prob = model.predict_proba(features)[0]
    if pred == 1:
        print(f"🚨 PHISHING ({prob[1]*100:.1f}% risk): {url}")
    else:
        print(f"✅ SAFE ({prob[0]*100:.1f}% safe): {url}")
```
